# **[3] Cleaning Data**

```python
df = df.dropna(how='all')
df.shape
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%m/%d/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%m/%d/%Y')

df[['Order Date', 'Ship Date']].dtypes
```

```python
df['Order Year'] = df['Order Date'].dt.year
```

```python
df[['Order Date', 'Order Year']].head()
```

```python
df['Order Month'] = df['Order Date'].dt.month
```

```python
df[['Order Date', 'Order Month', 'Order Year']].head()
```

```python
df['Shipping Duration'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
df[['Order Date', 'Ship Date', 'Shipping Duration']].head()
```

```python
df['Profit Margin (%)'] = (df['Profit'] / df['Sales']) * 100
```

```python
df[['Sales', 'Profit', 'Profit Margin (%)']].head()
```

```python
df['Order Quarter'] = df['Order Date'].dt.quarter
```

```python
df[['Order Date', 'Order Month', 'Order Year', 'Order Quarter']].head()
```

```python
df['Order Month Name'] = df['Order Date'].dt.month_name()
```

```python
df['Profit Category'] = df['Profit Margin (%)'].apply(lambda x: 'High' if x > 30 else ('Low' if x < 0 else 'Medium'))
```

```python
df['Return Flag'] = df['Profit'] < 0
```

```python
def categorize_shipping(days):
    if days == 0:
        return 'Same Day'
    elif 1 <= days <= 2:
        return 'Fast'
    elif 3 <= days <= 5:
        return 'Standard'
    else:
        return 'Delayed'

df['Shipping Category'] = df['Shipping Duration'].apply(categorize_shipping)
```

```python
df.head()
```

```python
new_column_order = [
    # المعرّفات
    'Row ID',

    # معلومات الطلب
    'Order ID', 'Order Date', 'Order Year', 'Order Quarter', 'Order Month', 'Order Month Name',
    'Ship Date', 'Shipping Duration', 'Shipping Category', 'Ship Mode',

    # معلومات العميل
    'Customer ID', 'Customer Name', 'Segment', 'Country', 'City', 'State', 'Postal Code', 'Region',

    # معلومات المنتج
    'Product ID', 'Product Name', 'Category', 'Sub-Category',

    # المقاييس المالية
    'Sales', 'Quantity', 'Discount', 'Profit', 'Profit Margin (%)', 'Profit Category', 'Return Flag'
]
```

```python
df = df[new_column_order]
df.head()
```

```python
sales_profit_by_year = df.groupby('Order Year')[['Sales', 'Profit']].sum()
sales_profit_by_year
```

```python
profit_by_category = df.groupby('Category')['Profit'].sum().sort_values(ascending=False)
profit_by_category
```

```python
losses_by_category = df[df['Return Flag'] == True].groupby('Category').size()
losses_by_category
```

```python
# إجمالي عدد الصفوف في كل فئة
total_by_category = df['Category'].value_counts()

# عدد الخسائر في كل فئة (لديكها بالفعل)
losses_by_category = df[df['Return Flag'] == True].groupby('Category').size()

# حساب نسبة الخسارة
loss_percentage = (losses_by_category / total_by_category) * 100
loss_percentage.sort_values(ascending=False)
```
